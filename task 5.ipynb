```python
# ============================================================
# TASK 5 ONLY — Downstream Evaluation from Frozen Encoder
# Clear, stable, VS Code local GPU version | One-cell notebook
# ============================================================

import os
import sys
import json
import time
import random
import subprocess
import importlib.util
from pathlib import Path
from dataclasses import dataclass
from typing import Dict, Any

# ------------------------------------------------------------
# 0) Install missing packages if needed
# ------------------------------------------------------------
def ensure_package(pkg_name: str, pip_name: str = None):
    pip_name = pip_name or pkg_name
    if importlib.util.find_spec(pkg_name) is None:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pip_name])

ensure_package("timm", "timm==1.0.9")
ensure_package("sklearn", "scikit-learn==1.4.2")
ensure_package("fvcore", "fvcore==0.1.5.post20221221")

# ------------------------------------------------------------
# 1) Imports
# ------------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import timm

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    roc_auc_score,
    roc_curve
)
from sklearn.preprocessing import label_binarize
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

try:
    from fvcore.nn import FlopCountAnalysis
    FVCORE_OK = True
except:
    FVCORE_OK = False
    print("fvcore not available. GFLOPs will be skipped.")

# ------------------------------------------------------------
# 2) Config
# ------------------------------------------------------------
@dataclass
class CFG:
    DATA_DIR: str = r"C:\Users\User\Desktop\Brain Cancer"
    OUT_DIR: str = r"D:\Rakib\final\Outputs_task4_ssl1_simclr"
    FROZEN_ENCODER_PATH: str = r"D:\Rakib\final\Outputs_task4_ssl1_simclr\simclr_frozen_encoder.pth"

    seed: int = 42

    # same split style as Task 4
    train_pct: int = 90
    test_pct: int = 10
    val_in_train: float = 0.10

    # safe local settings
    img_size: int = 160
    batch_size_sup: int = 16
    num_workers: int = 0

    # MUST match Task 4 backbone
    backbone: str = "resnet18"

    # feature-head training
    probe_epochs: int = 10
    probe_lr: float = 1e-3

    mlp_epochs: int = 10
    mlp_lr: float = 1e-3

    # full fine-tuning
    finetune_epochs: int = 8
    finetune_lr: float = 1e-4

    weight_decay: float = 1e-4
    deterministic: bool = True

cfg = CFG()
os.makedirs(cfg.OUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# 3) Device check
# ------------------------------------------------------------
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
AMP = torch.cuda.is_available()
PIN_MEMORY = torch.cuda.is_available()

print("=" * 72)
print("PyTorch version :", torch.__version__)
print("CUDA available  :", torch.cuda.is_available())
print("Selected device :", DEVICE)
if torch.cuda.is_available():
    print("GPU name        :", torch.cuda.get_device_name(0))
    print("GPU count       :", torch.cuda.device_count())
print("Output folder   :", cfg.OUT_DIR)
print("=" * 72)

# ------------------------------------------------------------
# 4) Reproducibility
# ------------------------------------------------------------
def seed_everything(seed: int, deterministic: bool = True):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    if deterministic:
        torch.backends.cudnn.deterministic = True
        torch.backends.cudnn.benchmark = False
    else:
        torch.backends.cudnn.deterministic = False
        torch.backends.cudnn.benchmark = True

seed_everything(cfg.seed, cfg.deterministic)

# ------------------------------------------------------------
# 5) Path checks
# ------------------------------------------------------------
assert os.path.exists(cfg.DATA_DIR), f"DATA_DIR not found:\n{cfg.DATA_DIR}"
assert os.path.exists(cfg.FROZEN_ENCODER_PATH), (
    f"Frozen encoder not found:\n{cfg.FROZEN_ENCODER_PATH}\n"
    f"Run Task 4 first or fix the path."
)

# ------------------------------------------------------------
# 6) Scan dataset
# ------------------------------------------------------------
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}

def scan_dataset(root: str) -> pd.DataFrame:
    rows = []
    root = Path(root)
    for p in root.rglob("*"):
        if p.is_file() and p.suffix.lower() in IMG_EXTS:
            rows.append((str(p), p.parent.name))
    return pd.DataFrame(rows, columns=["path", "label_name"])

df = scan_dataset(cfg.DATA_DIR)
assert len(df) > 0, f"No images found under {cfg.DATA_DIR}"

label_names = sorted(df["label_name"].unique().tolist())
label_to_idx = {name: i for i, name in enumerate(label_names)}
idx_to_label = {i: name for name, i in label_to_idx.items()}
df["label"] = df["label_name"].map(label_to_idx).astype(int)
num_classes = len(label_names)

print("Total images  :", len(df))
print("Total classes :", num_classes)
display(df.head())

# ------------------------------------------------------------
# 7) Split
# ------------------------------------------------------------
def make_split(df_all: pd.DataFrame, seed: int):
    test_ratio = cfg.test_pct / 100.0
    train_all, test_df = train_test_split(
        df_all,
        test_size=test_ratio,
        random_state=seed,
        stratify=df_all["label"]
    )
    val_df, train_df = train_test_split(
        train_all,
        test_size=(1 - cfg.val_in_train),
        random_state=seed,
        stratify=train_all["label"]
    )
    return train_df.reset_index(drop=True), val_df.reset_index(drop=True), test_df.reset_index(drop=True)

train_df, val_df, test_df = make_split(df, cfg.seed)

print("Train size:", len(train_df))
print("Val size  :", len(val_df))
print("Test size :", len(test_df))

with open(os.path.join(cfg.OUT_DIR, "task5_split_manifest.json"), "w", encoding="utf-8") as f:
    json.dump({
        "seed": cfg.seed,
        "train_pct": cfg.train_pct,
        "test_pct": cfg.test_pct,
        "val_in_train": cfg.val_in_train,
        "label_to_idx": label_to_idx,
        "train_paths": train_df["path"].tolist(),
        "val_paths": val_df["path"].tolist(),
        "test_paths": test_df["path"].tolist()
    }, f, indent=2)

# ------------------------------------------------------------
# 8) Transforms
# ------------------------------------------------------------
normalize = transforms.Normalize(
    mean=(0.485, 0.456, 0.406),
    std=(0.229, 0.224, 0.225)
)

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(cfg.img_size, scale=(0.7, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    normalize
])

eval_transform = transforms.Compose([
    transforms.Resize(int(cfg.img_size * 1.15)),
    transforms.CenterCrop(cfg.img_size),
    transforms.ToTensor(),
    normalize
])

# ------------------------------------------------------------
# 9) Dataset / loaders
# ------------------------------------------------------------
class ImgDataset(Dataset):
    def __init__(self, df: pd.DataFrame, transform):
        self.paths = df["path"].tolist()
        self.labels = df["label"].tolist()
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        p = self.paths[idx]
        y = self.labels[idx]
        img = Image.open(p).convert("RGB")
        x = self.transform(img)
        return x, y, p

sup_train_ds = ImgDataset(train_df, train_transform)
sup_val_ds = ImgDataset(val_df, eval_transform)
sup_test_ds = ImgDataset(test_df, eval_transform)

feat_train_ds = ImgDataset(train_df, eval_transform)
feat_val_ds = ImgDataset(val_df, eval_transform)
feat_test_ds = ImgDataset(test_df, eval_transform)

sup_train_loader = DataLoader(
    sup_train_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=True,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

sup_val_loader = DataLoader(
    sup_val_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=False,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

sup_test_loader = DataLoader(
    sup_test_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=False,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

feat_train_loader = DataLoader(
    feat_train_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=False,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

feat_val_loader = DataLoader(
    feat_val_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=False,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

feat_test_loader = DataLoader(
    feat_test_ds,
    batch_size=cfg.batch_size_sup,
    shuffle=False,
    num_workers=cfg.num_workers,
    pin_memory=PIN_MEMORY
)

# ------------------------------------------------------------
# 10) Encoder loader
# ------------------------------------------------------------
def get_encoder(backbone_name: str):
    encoder = timm.create_model(
        backbone_name,
        pretrained=False,
        num_classes=0,
        global_pool="avg"
    )
    feat_dim = encoder.num_features
    return encoder, feat_dim

frozen_encoder, feat_dim = get_encoder(cfg.backbone)
frozen_encoder.load_state_dict(torch.load(cfg.FROZEN_ENCODER_PATH, map_location="cpu"))
frozen_encoder = frozen_encoder.to(DEVICE).eval()

print("Frozen encoder loaded successfully.")
print("Feature dimension:", feat_dim)

# sanity check
x_chk, y_chk, p_chk = next(iter(feat_train_loader))
with torch.no_grad():
    h_chk = frozen_encoder(x_chk.to(DEVICE))
print("Feature sanity check:", h_chk.shape)

# ------------------------------------------------------------
# 11) Feature extraction
# ------------------------------------------------------------
@torch.no_grad()
def extract_features(encoder, loader):
    encoder.eval()
    feats, ys, paths = [], [], []
    for x, y, p in loader:
        x = x.to(DEVICE, non_blocking=True)
        h = encoder(x)
        feats.append(h.cpu().numpy())
        ys.append(y.numpy())
        paths.extend(list(p))
    return np.concatenate(feats, axis=0), np.concatenate(ys, axis=0), paths

print("\nExtracting train/val/test features...")
Xtr, ytr, _ = extract_features(frozen_encoder, feat_train_loader)
Xva, yva, _ = extract_features(frozen_encoder, feat_val_loader)
Xte, yte, _ = extract_features(frozen_encoder, feat_test_loader)

print("Train features:", Xtr.shape)
print("Val features  :", Xva.shape)
print("Test features :", Xte.shape)

# ------------------------------------------------------------
# 12) Metrics helpers
# ------------------------------------------------------------
def to_jsonable(obj):
    if isinstance(obj, dict):
        return {k: to_jsonable(v) for k, v in obj.items()}
    if isinstance(obj, list):
        return [to_jsonable(v) for v in obj]
    if isinstance(obj, tuple):
        return [to_jsonable(v) for v in obj]
    if isinstance(obj, np.integer):
        return int(obj)
    if isinstance(obj, np.floating):
        return float(obj)
    if isinstance(obj, np.ndarray):
        return obj.tolist()
    return obj

def per_class_accuracy(y_true, y_pred, n_classes: int):
    out = {}
    for c in range(n_classes):
        mask = (y_true == c)
        out[str(c)] = float((y_pred[mask] == y_true[mask]).mean()) if mask.sum() > 0 else float("nan")
    return out

def compute_metrics(y_true, y_pred, y_prob, n_classes: int):
    acc = float(accuracy_score(y_true, y_pred))
    p, r, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )

    y_true_bin = label_binarize(y_true, classes=list(range(n_classes)))
    auc = None
    try:
        auc = float(roc_auc_score(y_true_bin, y_prob, average="macro", multi_class="ovr"))
    except:
        auc = None

    return {
        "accuracy": acc,
        "precision_macro": float(p),
        "recall_macro": float(r),
        "f1_macro": float(f1),
        "roc_auc_ovr_macro": auc,
        "per_class_accuracy": per_class_accuracy(y_true, y_pred, n_classes)
    }

def plot_confmat(cm, class_names, title, save_path):
    plt.figure(figsize=(10, 8))
    plt.imshow(cm, interpolation="nearest")
    plt.title(title)
    plt.colorbar()
    ticks = np.arange(len(class_names))
    plt.xticks(ticks, class_names, rotation=90)
    plt.yticks(ticks, class_names)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.savefig(save_path, dpi=200)
    plt.show()

def plot_roc_curves(y_true, y_prob, n_classes, title, save_path, max_classes=10):
    y_true_bin = label_binarize(y_true, classes=list(range(n_classes)))
    plt.figure(figsize=(8, 6))
    plotted = 0
    for c in range(n_classes):
        if y_true_bin[:, c].sum() == 0:
            continue
        fpr, tpr, _ = roc_curve(y_true_bin[:, c], y_prob[:, c])
        plt.plot(fpr, tpr, label=f"class {c}")
        plotted += 1
        if plotted >= max_classes:
            break
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.title(title)
    plt.xlabel("FPR")
    plt.ylabel("TPR")
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.savefig(save_path, dpi=200)
    plt.show()

def top_confused_pairs(cm, class_names, topk=10):
    cm2 = cm.copy()
    np.fill_diagonal(cm2, 0)
    pairs = []
    for i in range(cm2.shape[0]):
        for j in range(cm2.shape[1]):
            if cm2[i, j] > 0:
                pairs.append((cm2[i, j], i, j))
    pairs.sort(reverse=True)
    return [{"true": class_names[i], "pred": class_names[j], "count": int(cnt)} for cnt, i, j in pairs[:topk]]

def save_json(path, obj):
    with open(path, "w", encoding="utf-8") as f:
        json.dump(to_jsonable(obj), f, indent=2)

# ------------------------------------------------------------
# 13) Feature-based models
# ------------------------------------------------------------
class LinearProbe(nn.Module):
    def __init__(self, in_dim, n_classes):
        super().__init__()
        self.fc = nn.Linear(in_dim, n_classes)

    def forward(self, x):
        return self.fc(x)

class MLPHead(nn.Module):
    def __init__(self, in_dim, n_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(256, n_classes)
        )

    def forward(self, x):
        return self.net(x)

def train_feature_model(model, X_train, y_train, X_val, y_val, epochs, lr, title):
    X_train_t = torch.tensor(X_train, dtype=torch.float32, device=DEVICE)
    y_train_t = torch.tensor(y_train, dtype=torch.long, device=DEVICE)
    X_val_t = torch.tensor(X_val, dtype=torch.float32, device=DEVICE)
    y_val_t = torch.tensor(y_val, dtype=torch.long, device=DEVICE)

    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    history = {"train_loss": [], "val_acc": []}

    best_acc = -1
    best_state = None
    start_time = time.time()

    for ep in range(epochs):
        model.train()
        perm = torch.randperm(X_train_t.size(0), device=DEVICE)
        batch_losses = []
        bs = 256

        for i in range(0, X_train_t.size(0), bs):
            idx = perm[i:i+bs]
            xb, yb = X_train_t[idx], y_train_t[idx]

            optimizer.zero_grad(set_to_none=True)
            logits = model(xb)
            loss = F.cross_entropy(logits, yb)
            loss.backward()
            optimizer.step()
            batch_losses.append(loss.item())

        model.eval()
        with torch.no_grad():
            val_logits = model(X_val_t)
            val_pred = val_logits.argmax(dim=1)
            val_acc = float((val_pred == y_val_t).float().mean().item())

        history["train_loss"].append(float(np.mean(batch_losses)))
        history["val_acc"].append(val_acc)

        if val_acc > best_acc:
            best_acc = val_acc
            best_state = {k: v.detach().cpu() for k, v in model.state_dict().items()}

        print(f"[{title}] Epoch {ep+1}/{epochs} | Train Loss: {np.mean(batch_losses):.4f} | Val Acc: {val_acc:.4f}")

    train_time = time.time() - start_time

    if best_state is not None:
        model.load_state_dict(best_state)

    return model, history, best_acc, train_time

def evaluate_feature_model(model, X_test, y_test):
    start = time.time()
    X_test_t = torch.tensor(X_test, dtype=torch.float32, device=DEVICE)
    with torch.no_grad():
        logits = model(X_test_t)
        prob = F.softmax(logits, dim=1).cpu().numpy()
        pred = logits.argmax(dim=1).cpu().numpy()
    test_time = time.time() - start
    metrics = compute_metrics(y_test, pred, prob, num_classes)
    cm = confusion_matrix(y_test, pred)
    return metrics, cm, pred, prob, test_time

# ------------------------------------------------------------
# 14) Linear Probe
# ------------------------------------------------------------
linear_probe = LinearProbe(feat_dim, num_classes).to(DEVICE)
linear_probe, probe_history, best_probe_val, probe_train_time = train_feature_model(
    linear_probe, Xtr, ytr, Xva, yva, cfg.probe_epochs, cfg.probe_lr, "Linear Probe"
)
probe_metrics, probe_cm, probe_pred, probe_prob, probe_test_time = evaluate_feature_model(
    linear_probe, Xte, yte
)

plt.figure(figsize=(7, 5))
plt.plot(probe_history["train_loss"], label="train_loss")
plt.plot(probe_history["val_acc"], label="val_acc")
plt.title("Linear Probe Learning Curve")
plt.xlabel("Epoch")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(cfg.OUT_DIR, "linear_probe_learning_curve.png"), dpi=200)
plt.show()

plot_confmat(probe_cm, label_names, "Linear Probe Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "linear_probe_confusion_matrix.png"))
plot_roc_curves(yte, probe_prob, num_classes, "Linear Probe ROC Curves",
                os.path.join(cfg.OUT_DIR, "linear_probe_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "linear_probe_metrics.json"), probe_metrics)

# ------------------------------------------------------------
# 15) MLP Head
# ------------------------------------------------------------
mlp_head = MLPHead(feat_dim, num_classes).to(DEVICE)
mlp_head, mlp_history, best_mlp_val, mlp_train_time = train_feature_model(
    mlp_head, Xtr, ytr, Xva, yva, cfg.mlp_epochs, cfg.mlp_lr, "MLP Head"
)
mlp_metrics, mlp_cm, mlp_pred, mlp_prob, mlp_test_time = evaluate_feature_model(
    mlp_head, Xte, yte
)

plt.figure(figsize=(7, 5))
plt.plot(mlp_history["train_loss"], label="train_loss")
plt.plot(mlp_history["val_acc"], label="val_acc")
plt.title("MLP Head Learning Curve")
plt.xlabel("Epoch")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(cfg.OUT_DIR, "mlp_head_learning_curve.png"), dpi=200)
plt.show()

plot_confmat(mlp_cm, label_names, "MLP Head Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "mlp_head_confusion_matrix.png"))
plot_roc_curves(yte, mlp_prob, num_classes, "MLP Head ROC Curves",
                os.path.join(cfg.OUT_DIR, "mlp_head_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "mlp_head_metrics.json"), mlp_metrics)

# ------------------------------------------------------------
# 16) SVM / Decision Tree / Random Forest
# These run on CPU. That is normal.
# ------------------------------------------------------------
svm_start = time.time()
svm = LinearSVC()
svm.fit(Xtr, ytr)
svm_train_time = time.time() - svm_start

svm_test_start = time.time()
svm_pred = svm.predict(Xte)
svm_test_time = time.time() - svm_test_start
svm_prob = np.eye(num_classes)[svm_pred]
svm_metrics = compute_metrics(yte, svm_pred, svm_prob, num_classes)
svm_cm = confusion_matrix(yte, svm_pred)

plot_confmat(svm_cm, label_names, "SVM Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "svm_confusion_matrix.png"))
plot_roc_curves(yte, svm_prob, num_classes, "SVM ROC Curves",
                os.path.join(cfg.OUT_DIR, "svm_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "svm_metrics.json"), svm_metrics)

dt_start = time.time()
dt = DecisionTreeClassifier(random_state=cfg.seed)
dt.fit(Xtr, ytr)
dt_train_time = time.time() - dt_start

dt_test_start = time.time()
dt_pred = dt.predict(Xte)
dt_prob = dt.predict_proba(Xte)
dt_test_time = time.time() - dt_test_start
dt_metrics = compute_metrics(yte, dt_pred, dt_prob, num_classes)
dt_cm = confusion_matrix(yte, dt_pred)

plot_confmat(dt_cm, label_names, "Decision Tree Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "decision_tree_confusion_matrix.png"))
plot_roc_curves(yte, dt_prob, num_classes, "Decision Tree ROC Curves",
                os.path.join(cfg.OUT_DIR, "decision_tree_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "decision_tree_metrics.json"), dt_metrics)

rf_start = time.time()
rf = RandomForestClassifier(n_estimators=300, random_state=cfg.seed, n_jobs=-1)
rf.fit(Xtr, ytr)
rf_train_time = time.time() - rf_start

rf_test_start = time.time()
rf_pred = rf.predict(Xte)
rf_prob = rf.predict_proba(Xte)
rf_test_time = time.time() - rf_test_start
rf_metrics = compute_metrics(yte, rf_pred, rf_prob, num_classes)
rf_cm = confusion_matrix(yte, rf_pred)

plot_confmat(rf_cm, label_names, "Random Forest Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "random_forest_confusion_matrix.png"))
plot_roc_curves(yte, rf_prob, num_classes, "Random Forest ROC Curves",
                os.path.join(cfg.OUT_DIR, "random_forest_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "random_forest_metrics.json"), rf_metrics)

# ------------------------------------------------------------
# 17) Full Fine-Tuning
# ------------------------------------------------------------
class FineTuneModel(nn.Module):
    def __init__(self, encoder, feat_dim, n_classes):
        super().__init__()
        self.encoder = encoder
        self.fc = nn.Linear(feat_dim, n_classes)

    def forward(self, x):
        h = self.encoder(x)
        return self.fc(h)

finetune_encoder, finetune_feat_dim = get_encoder(cfg.backbone)
finetune_encoder.load_state_dict(torch.load(cfg.FROZEN_ENCODER_PATH, map_location="cpu"))
finetune_model = FineTuneModel(finetune_encoder, finetune_feat_dim, num_classes).to(DEVICE)

ft_opt = torch.optim.AdamW(finetune_model.parameters(), lr=cfg.finetune_lr, weight_decay=cfg.weight_decay)
ft_sched = torch.optim.lr_scheduler.CosineAnnealingLR(ft_opt, T_max=cfg.finetune_epochs)
ft_scaler = torch.cuda.amp.GradScaler(enabled=AMP)

ft_history = {"train_loss": [], "val_loss": [], "train_acc": [], "val_acc": []}
best_ft_acc = -1
best_ft_state = None

ft_train_start = time.time()

for ep in range(cfg.finetune_epochs):
    finetune_model.train()
    tr_losses, tr_correct, tr_total = [], 0, 0

    for x, y, _ in sup_train_loader:
        x = x.to(DEVICE, non_blocking=True)
        y = y.to(DEVICE, non_blocking=True)

        ft_opt.zero_grad(set_to_none=True)

        with torch.cuda.amp.autocast(enabled=AMP):
            logits = finetune_model(x)
            loss = F.cross_entropy(logits, y)

        ft_scaler.scale(loss).backward()
        ft_scaler.step(ft_opt)
        ft_scaler.update()

        tr_losses.append(loss.item())
        pred = logits.argmax(dim=1)
        tr_correct += (pred == y).sum().item()
        tr_total += x.size(0)

    finetune_model.eval()
    va_losses, va_correct, va_total = [], 0, 0
    with torch.no_grad():
        for x, y, _ in sup_val_loader:
            x = x.to(DEVICE, non_blocking=True)
            y = y.to(DEVICE, non_blocking=True)

            logits = finetune_model(x)
            loss = F.cross_entropy(logits, y)

            va_losses.append(loss.item())
            pred = logits.argmax(dim=1)
            va_correct += (pred == y).sum().item()
            va_total += x.size(0)

    train_acc = tr_correct / tr_total if tr_total > 0 else 0.0
    val_acc = va_correct / va_total if va_total > 0 else 0.0

    ft_history["train_loss"].append(float(np.mean(tr_losses)) if tr_losses else 0.0)
    ft_history["val_loss"].append(float(np.mean(va_losses)) if va_losses else 0.0)
    ft_history["train_acc"].append(train_acc)
    ft_history["val_acc"].append(val_acc)

    if val_acc > best_ft_acc:
        best_ft_acc = val_acc
        best_ft_state = {k: v.detach().cpu() for k, v in finetune_model.state_dict().items()}

    ft_sched.step()
    print(f"[Fine-Tune] Epoch {ep+1}/{cfg.finetune_epochs} | Train Acc: {train_acc:.4f} | Val Acc: {val_acc:.4f}")

ft_train_time = time.time() - ft_train_start

if best_ft_state is not None:
    finetune_model.load_state_dict(best_ft_state)

ft_test_start = time.time()
ft_y_true, ft_y_pred, ft_y_prob = [], [], []

with torch.no_grad():
    for x, y, _ in sup_test_loader:
        x = x.to(DEVICE, non_blocking=True)
        logits = finetune_model(x)
        prob = F.softmax(logits, dim=1).cpu().numpy()
        pred = logits.argmax(dim=1).cpu().numpy()

        ft_y_true.append(y.numpy())
        ft_y_pred.append(pred)
        ft_y_prob.append(prob)

ft_test_time = time.time() - ft_test_start

ft_y_true = np.concatenate(ft_y_true, axis=0)
ft_y_pred = np.concatenate(ft_y_pred, axis=0)
ft_y_prob = np.concatenate(ft_y_prob, axis=0)

ft_metrics = compute_metrics(ft_y_true, ft_y_pred, ft_y_prob, num_classes)
ft_cm = confusion_matrix(ft_y_true, ft_y_pred)

plt.figure(figsize=(7, 5))
plt.plot(ft_history["train_loss"], label="train_loss")
plt.plot(ft_history["val_loss"], label="val_loss")
plt.plot(ft_history["train_acc"], label="train_acc")
plt.plot(ft_history["val_acc"], label="val_acc")
plt.title("Fine-Tuning Learning Curve")
plt.xlabel("Epoch")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(cfg.OUT_DIR, "finetune_learning_curve.png"), dpi=200)
plt.show()

plot_confmat(ft_cm, label_names, "Fine-Tune Confusion Matrix",
             os.path.join(cfg.OUT_DIR, "finetune_confusion_matrix.png"))
plot_roc_curves(ft_y_true, ft_y_prob, num_classes, "Fine-Tune ROC Curves",
                os.path.join(cfg.OUT_DIR, "finetune_roc_curves.png"))
save_json(os.path.join(cfg.OUT_DIR, "finetune_metrics.json"), ft_metrics)

# ------------------------------------------------------------
# 18) GFLOPs
# ------------------------------------------------------------
def estimate_gflops(model, img_size):
    if not FVCORE_OK:
        return None
    model = model.to(DEVICE).eval()
    x = torch.randn(1, 3, img_size, img_size, device=DEVICE)
    flops = FlopCountAnalysis(model, x).total()
    return float(flops) / 1e9

probe_wrapper = nn.Sequential(
    frozen_encoder.to(DEVICE),
    nn.Linear(feat_dim, num_classes).to(DEVICE)
).to(DEVICE)

gflops_probe = estimate_gflops(probe_wrapper, cfg.img_size)
gflops_finetune = estimate_gflops(finetune_model, cfg.img_size)

# ------------------------------------------------------------
# 19) Final summary / results table
# ------------------------------------------------------------
summary = {
    "task": "Task 5",
    "ssl_source": "SimCLR frozen encoder from Task 4",
    "backbone": cfg.backbone,
    "img_size": cfg.img_size,
    "device": str(DEVICE),
    "gpu_name": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None,
    "used_local_gpu": bool(torch.cuda.is_available()),
    "split": f"{cfg.train_pct}:{cfg.test_pct}",

    "linear_probe": {
        "metrics": probe_metrics,
        "train_time_sec": probe_train_time,
        "test_time_sec": probe_test_time,
        "failure_modes": top_confused_pairs(probe_cm, label_names, topk=10),
    },
    "mlp_head": {
        "metrics": mlp_metrics,
        "train_time_sec": mlp_train_time,
        "test_time_sec": mlp_test_time,
        "failure_modes": top_confused_pairs(mlp_cm, label_names, topk=10),
    },
    "svm": {
        "metrics": svm_metrics,
        "train_time_sec": svm_train_time,
        "test_time_sec": svm_test_time,
        "failure_modes": top_confused_pairs(svm_cm, label_names, topk=10),
    },
    "decision_tree": {
        "metrics": dt_metrics,
        "train_time_sec": dt_train_time,
        "test_time_sec": dt_test_time,
        "failure_modes": top_confused_pairs(dt_cm, label_names, topk=10),
    },
    "random_forest": {
        "metrics": rf_metrics,
        "train_time_sec": rf_train_time,
        "test_time_sec": rf_test_time,
        "failure_modes": top_confused_pairs(rf_cm, label_names, topk=10),
    },
    "full_finetune": {
        "metrics": ft_metrics,
        "train_time_sec": ft_train_time,
        "test_time_sec": ft_test_time,
        "failure_modes": top_confused_pairs(ft_cm, label_names, topk=10),
    },
    "gflops_linear_probe_wrapper": gflops_probe,
    "gflops_finetuned_model": gflops_finetune,
}

save_json(os.path.join(cfg.OUT_DIR, "task5_summary.json"), summary)

results_table = pd.DataFrame([
    {
        "Method": "Linear Probe",
        "Accuracy": probe_metrics["accuracy"],
        "Precision": probe_metrics["precision_macro"],
        "Recall": probe_metrics["recall_macro"],
        "F1": probe_metrics["f1_macro"],
        "ROC-AUC": probe_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": probe_train_time,
        "Test Time (s)": probe_test_time,
        "GFLOPs": gflops_probe,
    },
    {
        "Method": "MLP Head",
        "Accuracy": mlp_metrics["accuracy"],
        "Precision": mlp_metrics["precision_macro"],
        "Recall": mlp_metrics["recall_macro"],
        "F1": mlp_metrics["f1_macro"],
        "ROC-AUC": mlp_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": mlp_train_time,
        "Test Time (s)": mlp_test_time,
        "GFLOPs": gflops_probe,
    },
    {
        "Method": "SVM",
        "Accuracy": svm_metrics["accuracy"],
        "Precision": svm_metrics["precision_macro"],
        "Recall": svm_metrics["recall_macro"],
        "F1": svm_metrics["f1_macro"],
        "ROC-AUC": svm_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": svm_train_time,
        "Test Time (s)": svm_test_time,
        "GFLOPs": None,
    },
    {
        "Method": "Decision Tree",
        "Accuracy": dt_metrics["accuracy"],
        "Precision": dt_metrics["precision_macro"],
        "Recall": dt_metrics["recall_macro"],
        "F1": dt_metrics["f1_macro"],
        "ROC-AUC": dt_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": dt_train_time,
        "Test Time (s)": dt_test_time,
        "GFLOPs": None,
    },
    {
        "Method": "Random Forest",
        "Accuracy": rf_metrics["accuracy"],
        "Precision": rf_metrics["precision_macro"],
        "Recall": rf_metrics["recall_macro"],
        "F1": rf_metrics["f1_macro"],
        "ROC-AUC": rf_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": rf_train_time,
        "Test Time (s)": rf_test_time,
        "GFLOPs": None,
    },
    {
        "Method": "Full Fine-Tuning",
        "Accuracy": ft_metrics["accuracy"],
        "Precision": ft_metrics["precision_macro"],
        "Recall": ft_metrics["recall_macro"],
        "F1": ft_metrics["f1_macro"],
        "ROC-AUC": ft_metrics["roc_auc_ovr_macro"],
        "Train Time (s)": ft_train_time,
        "Test Time (s)": ft_test_time,
        "GFLOPs": gflops_finetune,
    },
])

for c in ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC", "Train Time (s)", "Test Time (s)"]:
    results_table[c] = pd.to_numeric(results_table[c], errors="coerce").round(4)

results_table.to_csv(os.path.join(cfg.OUT_DIR, "task5_results_table.csv"), index=False)

print("\n================ TASK 5 COMPLETE ================\n")
print("Used local GPU:", torch.cuda.is_available())
print("Summary saved :", os.path.join(cfg.OUT_DIR, "task5_summary.json"))
print("Table saved   :", os.path.join(cfg.OUT_DIR, "task5_results_table.csv"))
display(results_table.sort_values("Accuracy", ascending=False).reset_index(drop=True))
```
